# GemelNet Data Scraper

Comprehensive scraper for Israel's Capital Market Authority (CMA) GemelNet system.

**Data sources:**
- GemelNet XML API (`tsuotHodXML.aspx`) - Monthly portfolio & performance data
- Israel Open Data Portal (data.gov.il) - CKAN API dataset
- GemelNet website pages - ASP.NET pages with fund information
- PensiaNet - Related CMA pension system

**Output:** `./gemelnet_data/`

```python
# Step 3: Explore downloaded data
import os
import pandas as pd

data_dir = 'gemelnet_data'
print('Downloaded files:')
print('=' * 60)
for root, dirs, files in os.walk(data_dir):
    level = root.replace(data_dir, '').count(os.sep)
    indent = ' ' * 2 * level
    print(f'{indent}{os.path.basename(root)}/')
    subindent = ' ' * 2 * (level + 1)
    for f in sorted(files):
        path = os.path.join(root, f)
        size = os.path.getsize(path) / 1024
        print(f'{subindent}{f} ({size:.1f} KB)')
```

```python
# Step 4: Load and inspect CSV data
import glob

csv_files = glob.glob('gemelnet_data/**/*.csv', recursive=True)
print(f'Found {len(csv_files)} CSV files\n')

for csv_file in csv_files[:5]:  # Show first 5
    print(f'\n--- {csv_file} ---')
    try:
        df = pd.read_csv(csv_file, encoding='utf-8-sig', nrows=5)
        print(f'Columns: {list(df.columns)}')
        print(f'Shape: {df.shape}')
        display(df.head())
    except Exception as e:
        print(f'Error reading: {e}')
```
